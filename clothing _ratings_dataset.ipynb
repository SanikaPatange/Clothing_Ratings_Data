```python
df2.loc[df2['Rating'].isin([1,2]),"Sentiment"] = "Negative" #Uses the 'loc' to select and modify specific rows and columns
df2.loc[df2['Rating'].isin([4,5]),"Sentiment"] = "Positive"
```

```python
#Train - Test

from sklearn.model_selection import train_test_split
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LinearRegression
```

```python
x = df2['Review Text']
y = df2['Sentiment']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.3 , random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
vec = CountVectorizer() # CountVectorizer use to convert text into ML language and then count it. Example : good quality = [1,0,1,0,1,0,0]
```

```python
x_train = x_train.fillna('')  # Replace NaN/Null with empty string
x_test = x_test.fillna('')
```

```python
x_train_vec = vec.fit_transform(x_train)
x_test_vec = vec.transform(x_test)
```

```python
x_train_vec.shape
```

```python
x_test_vec.shape
```

```python
from sklearn.naive_bayes import MultinomialNB 
# MultinominalNB is used to count the word in one sentense. Example: good good quality - good = 2, quality = 1
```

```python
model = MultinomialNB()
```

```python
model.fit(x_train_vec, y_train)
```

```python
y_pred = model.predict(x_test_vec)
```

```python
acc = accuracy_score(y_test, y_pred)
```

```python
confusion_matrix(y_test, y_pred)
```

```python
print(classification_report(y_test, y_pred))
```

- **Accurac**y is 81%
- **Precision** : It treats that when our model predicts "Negative sentiment".how often it is actually "Negative Sentiment".
- **recall** : Out of all the negative Sentiment , how many did our model successfully catch.
- **f1-Score** : If precision is high but recall is poor - F1 goes down. balance of both
  - It is used to identify when classes are imbalanced .
- **classification report** : complete reeport of model.
- **confusion matrix** : its a error table which shows how many tp, tn, fp, fn are there.
